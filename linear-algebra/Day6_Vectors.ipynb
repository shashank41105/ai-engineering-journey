```python
import numpy as np
def check_span(v1,v2,w):
    """
    determines if vector w is a linear combination of v1 and v2
    returns (true, scalars) if reachable ,(false,none) otherwise
    """
    #step1: create a matrix A where v1 and v2 are columns
    #this represents the "available directions " in our space
    A=np.column_stack((v1,v2))
    try:
        #step2: solve the equation Ax=w
        # x will be our scalars[a,b]
        scalars=np.linalg.solve(A,w)
        return True,scalars
    except np.linalg.LinAlgError:
        # if the vectors are collinear (linearly dependent),
        # the matrix is "singular" and cannot be inverted 
        return False,None
#test cases:
#case1: reachable ( the standard basis)
v1= np.array([1,0])
v2=np.array([0,1])
w1=np.array([3,4]) #expected : true,scalars=[3,4]
print(f" case1:{check_span(v1,v2,w1)}")
#case2: unreachable (collinear vectors)
v1_col=np.array([1,2])
v2_col=np.array([2,4]) # this is just 2*v1_col
w2=np.array([1,1]) # this point is Not on the line
print(f"case2: {check_span(v1_col,v2_col,w2)}")
#case3: reachable but collinear (special case)
w3=np.array([3,6]) #this point IS on the line [1,2]
#note: mp.linalg.solve might still fail here because the matrix 
#is singular ,even if a solution exists. This is a great talking poit!
print(f" case3: {check_span(v1_col,v2_col,w3)}")

    
```
