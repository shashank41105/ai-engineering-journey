```python
import numpy as np
import matplotlib.pyplot as plt

def plot_vectors(vectors, colors, title):
    plt.figure(figsize=(5, 5))
    plt.axhline(0, color='black', lw=1)
    plt.axvline(0, color='black', lw=1)
    
    # Plot each vector as an arrow
    for i in range(len(vectors)):
        plt.quiver(0, 0, vectors[i][0], vectors[i][1], 
                   angles='xy', scale_units='xy', scale=1, color=colors[i])
    
    plt.xlim(-5, 5)
    plt.ylim(-5, 5)
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.title(title)
    plt.show()

# 1. Define our original vectors (The "Before" state)
# Let's use the basis vectors i-hat (1,0) and j-hat (0,1) 
# and one random vector (1,1)
original_vectors = np.array([
    [1, 0], # i-hat
    [0, 1], # j-hat
    [1, 1]  # v
])

# 2. Define the Transformation Matrix (A)
# Let's try a 90-degree counter-clockwise rotation matrix:
# [ 0  -1 ]
# [ 1   0 ]
A = np.array([
    [0, -1],
    [1,  0]
])

# 3. Perform Matrix-Vector Multiplication
# We transpose original_vectors to make them columns, multiply, then transpose back
transformed_vectors = (A @ original_vectors.T).T

# Visualization
colors = ['red', 'blue', 'green']
plot_vectors(original_vectors, colors, "Original Space (Basis vectors in Red/Blue)")
plot_vectors(transformed_vectors, colors, "Transformed Space (90° Rotation)")

print("Original Vectors:\n", original_vectors)
print("\nTransformed Vectors:\n", transformed_vectors)
```
