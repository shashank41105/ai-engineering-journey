```python
import numpy as np

# -----------------------------------------------------------------------------
# Level 1: The "Naive" Implementation
# -----------------------------------------------------------------------------
def manual_dot_product(v1, v2):
    """Calculates dot product using a basic for loop (no libraries)."""
    if len(v1) != len(v2):
        raise ValueError("Vectors must be of the same length")
    
    result = 0
    for i in range(len(v1)):
        result += v1[i] * v2[i]
    return result

# -----------------------------------------------------------------------------
# Level 2: The NumPy Power-up
# -----------------------------------------------------------------------------
def numpy_dot_product(v1, v2):
    """Calculates dot product using NumPy's optimized operators."""
    # We convert lists to numpy arrays first
    vec1 = np.array(v1)
    vec2 = np.array(v2)
    return np.dot(vec1, vec2)  # Or use vec1 @ vec2

# -----------------------------------------------------------------------------
# Level 3: The Geometric Tool (Angle Calculation)
# -----------------------------------------------------------------------------
def calculate_angle(v1, v2):
    """Returns the angle between two vectors in degrees."""
    vec1 = np.array(v1)
    vec2 = np.array(v2)
    
    # Dot product
    dot_prod = np.dot(vec1, vec2)
    
    # Magnitudes (norms)
    norm_v1 = np.linalg.norm(vec1)
    norm_v2 = np.linalg.norm(vec2)
    
    # Cosine of the angle: cos(theta) = (a . b) / (|a| * |b|)
    cos_theta = dot_prod / (norm_v1 * norm_v2)
    
    # Clip cos_theta to be within [-1, 1] to avoid NaN from floating point errors
    cos_theta = np.clip(cos_theta, -1.0, 1.0)
    
    # Arccos to get radians, then convert to degrees
    radians = np.arccos(cos_theta)
    degrees = np.degrees(radians)
    
    return degrees

# =============================================================================
#  Testing the Implementation
# =============================================================================

test_cases = [
    {"name": "Parallel",   "v1": [1, 0], "v2": [2, 0]},
    {"name": "Orthogonal", "v1": [1, 0], "v2": [0, 1]},
    {"name": "Opposite",   "v1": [1, 0], "v2": [-1, 0]},
]

print(f"{'Case':<12} | {'Manual':<8} | {'NumPy':<8} | {'Angle (deg)':<12}")
print("-" * 45)

for case in test_cases:
    m_res = manual_dot_product(case["v1"], case["v2"])
    n_res = numpy_dot_product(case["v1"], case["v2"])
    angle = calculate_angle(case["v1"], case["v2"])
    
    print(f"{case['name']:<12} | {m_res:<8} | {n_res:<8} | {angle:<12.2f}")
```
