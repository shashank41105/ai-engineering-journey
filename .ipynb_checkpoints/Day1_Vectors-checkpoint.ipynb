```python

import numpy as np
a=np.array([2,3])
b=np.array([2,5])
sum_vec=a+b
print(f"Sum of a and b:{sum_vec}")
scaled_a=3*a
print(f"3 times of a:{scaled_a}")
c=np.array([1,2,-6])
d=np.array([-2,-2,-2])
print(f"length of vector a:{np.linalg.norm(a)}")
opp_b=b*-1
print(f"opposite of b:{opp_b}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
#1. Defining vectors 
v=np.array([2,3])
w=np.array([4,-1])
sum_vw=v+w
#2. Setup the plot
plt.figure(figsize=(6,6))
plt.axhline(0,color="black",linewidth=1) #X-axis
plt.axvline(0,color="black",linewidth=1)#Y-axis
plt.grid (True,linestyle="--",alpha=0.6)
#3. plot vectors using plt.quiver (this is the best way to draw arrows)
#origin:[0,0] ,vectors :[x,y]
plt.quiver([0,0,0],[0,0,0],[v[0],w[0],sum_vw[0]],[v[1],w[1],sum_vw[1]],
           angles='xy',scale_units='xy',scale=1,
           color=['r','b','g'],label=['v','w','v+w'])
#4. Final touches
plt.xlim(-1,7)
plt.ylim(-2,5)
plt.legend()
plt.title("vector addition visualization")
plt.show()
```

```python
import numpy as np 
import matplotlib.pyplot as plt
v=np.array([2,5])
u=2*v
print(u)
plt.figure(figsize=(6,6))
plt.axhline(0,color='black',linewidth=1)
plt.axvline(0,color='black',linewidth=1)
plt.grid(True,linestyle="--",alpha=0.6)
plt.quiver([0,0],[0,0],[v[0],u[0]],[v[1],u[1]],angles='xy',scale_units='xy',scale=1,
          color=['r','b'],label=['v','u'])
plt.xlim(-5,10)
plt.ylim(-6,15)
plt.legend(['v','u'])
plt.title("Scalar vector-Scaling")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
v=np.array([1,2])
w=np.array([3,1])
u=np.array([0,5])
sum_vwu=v+w+u
print(sum_vwu)
plt.figure(figsize=(10,10))
plt.axhline(0,color="black",linewidth=1)
plt.axvline(0,color='black',linewidth=1)
plt.grid(True,linestyle="--",alpha=0.6)
plt.quiver([0,0,0,0],[0,0,0,0],[v[0],w[0],u[0],sum_vwu[0]],[v[1],w[1],u[1],sum_vwu[1]],
           angles='xy',scale_units='xy',scale=1,
           color=['r','y','b','g'],label=['v','w','u','v+w+u'])
plt.xlim(-1,10)
plt.ylim(-1,10)
plt.legend()
plt.title("addition of 3 vectors")
plt.show()
```
